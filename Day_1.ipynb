```python
from google.colab import userdata
from openai import OpenAI

client = OpenAI(
    api_key=userdata.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

MODEL = "openai/gpt-oss-20b"

print("EduPredictAI API connected successfully!")
```

```python
def generate_tagline(temp):
    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {
                "role": "system",
                "content": "You are a creative marketing assistant for EduPredictAI, an AI-based student performance prediction project."
            },
            {
                "role": "user",
                "content": "Write a short tagline for EduPredictAI."
            }
        ],
        temperature=temp
    )

    print("Temperature:", temp)
    print(response.choices[0].message.content)
    print("-" * 50)


# Low temperature
for i in range(3):
    generate_tagline(0.2)

# High temperature
for i in range(3):
    generate_tagline(1.2)
```

```python
response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {
            "role": "system",
            "content": "You are an assistant for EduPredictAI, an AI-based student performance prediction project."
        },
        {
            "role": "user",
            "content": "Does EduPredictAI guarantee that every student will pass? Explain its success guarantee."
        }
    ],
    temperature=0.7
)

print(response.choices[0].message.content)
```

```python
# First API call
response1 = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": "My name is Ravi."}
    ]
)

print("First response:")
print(response1.choices[0].message.content)


# Second API call
response2 = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": "What is my name?"}
    ]
)

print("\nSecond response:")
print(response2.choices[0].message.content)
```

```python
response3 = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": "My name is Ravi."},
        {"role": "assistant", "content": "Nice to meet you, Ravi!"},
        {"role": "user", "content": "What is my name?"}
    ]
)

print(response3.choices[0].message.content)
```

```python
english_question = """
What is EduPredictAI and how does it help teachers
identify students who may fail?
"""

telugu_question = """
EduPredictAI అంటే ఏమిటి మరియు ఫెయిల్ అయ్యే అవకాశం
ఉన్న విద్యార్థులను గుర్తించడంలో ఉపాధ్యాయులకు
ఇది ఎలా సహాయపడుతుంది?
"""


# English API call
english_response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": english_question}
    ]
)

# Telugu API call
telugu_response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "user", "content": telugu_question}
    ]
)


print("English prompt tokens:",
      english_response.usage.prompt_tokens)

print("Telugu prompt tokens:",
      telugu_response.usage.prompt_tokens)
```

```python
SYSTEM = """
You are EduPredictAI Assistant.

PROJECT FACTS:
1. EduPredictAI is an AI-based student performance prediction project.
2. It uses historical student academic data and marks.
3. It predicts three outcomes: Pass, Fail, and Dropout.
4. It provides an interactive dashboard for educators.
5. It helps educators identify at-risk students and provide support.

RULES:
1. Answer in simple and understandable English.
2. Do not invent project features or unsupported information.
3. Never guarantee student results. Predictions are only guidance.

You must answer based on the above project facts.
"""

questions = [
    "What is EduPredictAI?",
    "How does EduPredictAI identify at-risk students?",
    "Can EduPredictAI guarantee that every student will pass?"
]

for question in questions:

    response = client.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": question}
        ],
        temperature=0.3
    )

    print("Question:", question)
    print("Answer:", response.choices[0].message.content)
    print("-" * 60)
```
