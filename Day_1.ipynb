```python
from google.colab import userdata
from openai import OpenAI

# Read API key from Colab Secrets
GROQ_API_KEY = userdata.get("GROQ_API_KEY")

# Create API client
client = OpenAI(
    api_key=GROQ_API_KEY,
    base_url="https://api.groq.com/openai/v1"
)

print("EduPredict AI API client created successfully!")
```

```python
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "system",
            "content": "You are EduPredict AI, an intelligent academic assistant."
        },
        {
            "role": "user",
            "content": "What is EduPredict AI and how does it help universities?"
        }
    ]
)

print(response.choices[0].message.content)
```

```python
SYSTEM = """
ROLE:
You are EduPredict AI, an intelligent university admission
and academic decision-support assistant.

PROJECT:
EduPredict AI is a full-stack Machine Learning-powered
university admission web application.

CORE MODULES:

1. GPA AND ACADEMIC PERFORMANCE PREDICTION:
Analyze historical exam scores, subject marks and attendance
to forecast expected semester or cumulative GPA.

2. ACADEMIC BRANCH RECOMMENDATION:
Analyze subject strengths, entrance examination metrics
and aptitude patterns to recommend suitable branches
such as CSE, AI/ML, ECE and Mechanical.

3. SCHOLARSHIP ELIGIBILITY:
Evaluate academic merit and financial or household
indicators to screen students for merit-based and
need-based scholarships.

4. ANALYTICS DASHBOARD:
Visualize GPA distributions, branch demand,
scholarship allocation and student performance trends.

TECHNOLOGY STACK:
Frontend: React.js, Vite, JavaScript, Tailwind CSS, Recharts.
Backend: Python and FastAPI.
Database: MongoDB.
Machine Learning: Scikit-learn.
Data Processing: Pandas and NumPy.

USER ROLES:
1. Administrator:
Manages students, staff, scholarship rules,
model monitoring and institution-wide reports.

2. Staff/Counselor:
Updates student records, runs predictions and
reviews scholarship recommendations for assigned students.

RULES:
- Answer only using the project information above.
- Never invent student marks, GPA predictions or results.
- Never guarantee scholarship approval.
- Do not claim that unimplemented features are already working.
- If information is missing, say:
  "This information is not currently available in the project details."
- Explain technical concepts in simple English.
- Keep answers under 100 words.
- Be friendly and professional.
"""

print("EduPredict AI system prompt created!")
```

```python
question = "Explain the four main modules of EduPredict AI."

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": question}
    ]
)

answer = response.choices[0].message.content

print("EduPredict AI Assistant:")
print(answer)
```

```python
question = "Explain why EduPredict AI is useful for universities."

for temp in [0.2, 1.2]:

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        temperature=temp,
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": question}
        ]
    )

    print("\nTemperature:", temp)
    print(response.choices[0].message.content)
```

```python
question = "What technologies are used in EduPredict AI?"

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": question}
    ]
)

print("Answer:")
print(response.choices[0].message.content)

print("\nToken Usage:")
print("Prompt Tokens:", response.usage.prompt_tokens)
print("Completion Tokens:", response.usage.completion_tokens)
print("Total Tokens:", response.usage.total_tokens)
```

```python
def edupredict_assistant(question):

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        temperature=0.2,
        messages=[
            {"role": "system", "content": SYSTEM},
            {"role": "user", "content": question}
        ]
    )

    print("\n🎓 EduPredict AI Assistant")
    print("-" * 40)
    print(response.choices[0].message.content)
    print("-" * 40)

    print("Tokens Used:", response.usage.total_tokens)


# Test questions
edupredict_assistant("How does GPA prediction work?")

edupredict_assistant("How does branch recommendation help students?")

edupredict_assistant("What is the difference between merit and need-based scholarships?")

edupredict_assistant("Explain the technology stack.")
```
