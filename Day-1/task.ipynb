```python

!pip install -q openai

from google.colab import userdata
from openai import OpenAI

client = OpenAI(
    api_key=userdata.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1"
)

print("Groq API client created successfully!")
```

```python

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": "You are a helpful AI assistant."},
        {"role": "user", "content": "Explain Artificial Intelligence in simple English."}
    ]
)

print(response.choices[0].message.content)
```

```python

SYSTEM = """
ROLE:
You are the friendly AI assistant for EduPredictAI.

FACTS:
1. EduPredictAI is an AI-powered student academic performance prediction project.
2. It uses historical student academic data.
3. It classifies student outcomes as Pass, Fail, or Dropout.
4. It helps teachers identify students who may need academic support.
5. It uses Python, Pandas, NumPy and Machine Learning.

RULES:
1. Never invent student marks or academic information.
2. Never guarantee a student's future academic result.
3. Ask for missing information when necessary.

FORMAT:
Use simple English.
Give short, clear and helpful answers.
"""

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    temperature=0.2,
    messages=[
        {"role": "system", "content": SYSTEM},
        {"role": "user", "content": "How does EduPredictAI help teachers?"}
    ]
)

print(response.choices[0].message.content)
print("\nToken Usage:", response.usage)
```

```python

questions = [
    "Explain how EduPredictAI helps teachers identify students who need academic support.",
    "విద్యార్థులకు విద్యా సహాయం అవసరమని గుర్తించడంలో EduPredictAI ఉపాధ్యాయులకు ఎలా సహాయపడుతుంది?"
]

for question in questions:

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        messages=[
            {"role": "user", "content": question}
        ]
    )

    print("\nQuestion:", question)
    print("Answer:", response.choices[0].message.content)
    print("Prompt tokens:", response.usage.prompt_tokens)
    print("Completion tokens:", response.usage.completion_tokens)
```

```python

EDUPREDICTAI_SYSTEM = """
ROLE:
You are EduPredictAI's student academic support assistant.

FACTS:
1. EduPredictAI predicts student academic performance.
2. It uses historical academic data.
3. It classifies outcomes as Pass, Fail, or Dropout.
4. It helps teachers identify students who need support.
5. It uses Python and Machine Learning.

RULES:
1. Never invent student data.
2. Never guarantee predictions.
3. If information is missing, ask the user.

FORMAT:
Use simple English and short answers.
"""

questions = [
    "What is EduPredictAI?",
    "How can teachers use EduPredictAI?",
    "What are the three student performance categories?"
]

for question in questions:

    response = client.chat.completions.create(
        model="openai/gpt-oss-20b",
        temperature=0.2,
        messages=[
            {"role": "system", "content": EDUPREDICTAI_SYSTEM},
            {"role": "user", "content": question}
        ]
    )

    print("\nQuestion:", question)
    print("Answer:", response.choices[0].message.content)
    print("Token usage:", response.usage)
```
